```python
#replacing USD with 1
df['invoice_currency']=[x.replace('USD',"1") for x in df['invoice_currency']]
#replacing CAD with 0.81
df['invoice_currency']=[x.replace('CAD',"0.81") for x in df['invoice_currency']]

df['invoice_currency'] = df['invoice_currency'].astype(float,errors='raise')

#multiplying the columns 'total_open_amount' and 'invoice_currency' to convert CAD into USD so that amounts are calculated
#in same currency and storing it again to the 'total_open_amount' column
df['total_open_amount'] = df['total_open_amount'] * df['invoice_currency']
```

```python
df['total_open_amount']=df['total_open_amount'].astype(int)
df
```

```python
# function will delete a column from train , val and test
def del_unwanted(col ,traindf = X_train,valdf = X_val,testdf = X_test):
    traindf.drop(col, axis =1,inplace=True)
    valdf.drop(col,axis=1 , inplace=True)
    testdf.drop(col,axis=1 , inplace=True)

    return traindf,valdf ,testdf
```

```python
#bucketisation function

def bucketisation(value):
    if value > 60:
        return ">60 days"
    elif value > 45:
        return "46-60 days"
    elif value > 30:
        return "31-45 days"
    elif value > 15:
        return "16-30 days"
    elif value >=0:
        return "0-15 days"
```

```python
# bucketisation of data


X_train["Aging_Bucket"] = y_train.apply(bucketisation)
X_val["Aging_Bucket"] = y_val.apply(bucketisation)
X_test["Aging_Bucket"] = y_test.apply(bucketisation)
```
